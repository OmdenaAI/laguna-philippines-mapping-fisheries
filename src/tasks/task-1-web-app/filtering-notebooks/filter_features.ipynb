```python
import json
import requests
```

```python
geojson_url = 'https://raw.githubusercontent.com/OmdenaAI/laguna-philippines-mapping-fisheries/main/src/data/Results/region2_results_gdf_with_forecast_and_recommendation.geojson'
response = requests.get(geojson_url)
geojson_data = response.json()
# print(geojson_data)
```

```python
geojson_data
```

```python
features_keep = ["index","Recommendation", "Recommended Production Value (2022) - Forecasted", "Recommended Production Volume (2022) - Forecasted"]
```

```python
list(geojson_data["features"][0]['properties'])
```

```python
features_remove = list(geojson_data["features"][0]['properties'])
```

```python
for feature in features_keep:
    features_remove.remove(feature)
```

```python
for feature in features_remove:
    for i in range(len(geojson_data["features"])):
        geojson_data["features"][i]["properties"].pop(feature, None)
```

```python
geojson_data["features"][0]["properties"]
```

```python
# Save the geojson data to a file
with open('region_2_results_with_recommendation.geojson', 'w') as outfile:
    json.dump(geojson_data, outfile)
```
