```python
import json
import requests
```

```python
geojson_url = 'https://raw.githubusercontent.com/OmdenaAI/laguna-philippines-mapping-fisheries/main/src/data/Results/region_2_results.geojson'
response = requests.get(geojson_url)
geojson_data = response.json()
# print(geojson_data)
```

```python
features_keep = ["Production Value", "Production Volume", "Commercial Fisheries Volume", "Commercial Fisheries Value", "Marine Municipal Fisheries Volume", "Marine Municipal Fisheries Value", "Inland Municipal Fisheries Volume", "Inland Municipal Fisheries Value", "Aquaculture Volume", "Aquaculture Value"]
```

```python
features_remove = list(geojson_data["features"][1]["properties"])
```

```python
for feature in features_keep:
    # Use the "remove()" method to remove the current feature from the list
    features_remove.remove(feature)
```

```python
for feature in features_remove:
    for i in range(len(geojson_data["features"])):
        geojson_data["features"][i]["properties"].pop(feature, None)
```
